```python
# PostgreSQL Connection - Update these credentials with your actual database info
import os
from dotenv import load_dotenv

# Load environment variables from .env file (recommended for security)
load_dotenv()

db_host = os.getenv('DB_HOST', 'localhost')
db_user = os.getenv('DB_USER', 'postgres')
db_pass = os.getenv('DB_PASSWORD', 'your_password')  # Change this or use environment variable
db_name = os.getenv('DB_NAME', 'your_database')
db_port = os.getenv('DB_PORT', '5432')

# Create connection string
connection_string = f'postgresql+psycopg2://{db_user}:{db_pass}@{db_host}:{db_port}/{db_name}'

# Create SQLAlchemy engine
pg = create_engine(connection_string)

# Test connection
try:
    with pg.connect() as conn:
        result = conn.execute("SELECT 1")
        print("✓ Successfully connected to PostgreSQL!")
except Exception as e:
    print(f"✗ Connection failed: {e}")
    print("Please verify your database credentials in the environment variables or this cell")
```

```python
# Troubleshooting: Try common PostgreSQL default credentials
import subprocess

# First, check if PostgreSQL is running
try:
    result = subprocess.run(['pg_isready'], capture_output=True, text=True)
    print("PostgreSQL Status:", result.stdout.strip())
except:
    print("pg_isready not found - PostgreSQL might not be installed or in PATH")

# Try different password combinations
passwords_to_try = [
    '',  # No password
    'postgres',  # Default password
    'password',
    'admin',
    'root'
]

db_host = 'localhost'
db_user = 'postgres'
db_name = 'postgres'  # Default database

for pwd in passwords_to_try:
    connection_string = f'postgresql+psycopg2://{db_user}:{pwd}@{db_host}/{db_name}'
    try:
        engine = create_engine(connection_string)
        with engine.connect() as conn:
            conn.execute("SELECT 1")
        print(f"✓ SUCCESS! Password is: '{pwd}'")
        db_pass = pwd
        break
    except Exception as e:
        print(f"✗ Password '{pwd}' failed")
else:
    print("\nNone of the default passwords worked.")
    print("Please provide your PostgreSQL password manually.")
```

```python
# Find PostgreSQL Installation
import subprocess
import os
from pathlib import Path

# Method 1: Search where command
print("🔍 Searching for PostgreSQL installation...")

try:
    result = subprocess.run(['where', 'psql'], capture_output=True, text=True)
    if result.stdout.strip():
        psql_path = result.stdout.strip().split('\n')[0]
        print(f"✓ Found psql at: {psql_path}")
        
        # Get the bin directory
        bin_dir = os.path.dirname(psql_path)
        pg_root = os.path.dirname(bin_dir)
        
        print(f"✓ PostgreSQL root: {pg_root}")
        print(f"\n📝 YOUR POSTGRESQL INSTALLATION PATH:")
        print(f"   {pg_root}")
        
        # Try to find pg_ctl for password reset
        pg_ctl = os.path.join(bin_dir, 'pg_ctl.exe')
        if os.path.exists(pg_ctl):
            print(f"\n✓ Found pg_ctl at: {pg_ctl}")
    else:
        print("❌ psql not found in system PATH")
except Exception as e:
    print(f"Error: {e}")

# Method 2: Check environment variables
print("\n" + "="*60)
print("🔐 PASSWORD RECOVERY OPTIONS:")
print("="*60)

print("\n1️⃣  CHECK WINDOWS CREDENTIAL MANAGER:")
print("   - Press: Windows + R")
print("   - Type: credential")
print("   - Look for: 'postgresql' or 'localhost'")
print("   - Your password is stored there!")

print("\n2️⃣  RESET PASSWORD (requires admin access):")
print("   Open PowerShell as Administrator and run:")
print('   psql -U postgres -c "ALTER USER postgres WITH PASSWORD \'newpassword\';"')

print("\n3️⃣  RESET POSTGRES SERVICE (Nuclear Option):")
print("   - Windows Services → PostgreSQL")
print("   - Right-click → Restart")
print("   - Then try password: 'postgres' (factory default)")
```
